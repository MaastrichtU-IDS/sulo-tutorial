## OWL Primer ##

The **Web Ontology Language (OWL)** is a formal language for representing knowledge about a domain in a way that both humans and machines can understand. It is a core technology of the Semantic Web, designed to describe what exists in a domain, how things relate, and to enable automated reasoning over that knowledge.

An **OWL ontology** is a formal, machine-readable description of a domain of knowledge, written in OWL, that defines:

* what kinds of things exist (<span style="color:blue;font-weight:bold;">class</span>)
* which specific things exist (<span style="color:green;font-style:italic;">individuals</span>)
* how they are related (<span style="color:orange;">object properties</span>)
* and what (data) attributes they have (<span style="color:orange;">data properties</span>)

together with logical rules that constrain and structure this knowledge.

> **Color convention:** <span style="color:blue;font-weight:bold;">Classes</span> &nbsp;|&nbsp; <span style="color:green;font-style:italic;">Individuals</span> &nbsp;|&nbsp; <span style="color:orange;">Properties</span> &nbsp;|&nbsp; <span style="color:purple;font-family:monospace;">OWL keywords</span>

---

### Class ###
A <span style="color:blue;font-weight:bold;">Class</span> represents a category or type of thing in the world.
Examples: <span style="color:blue;font-weight:bold;">Person</span>, <span style="color:blue;font-weight:bold;">Pizza</span>, <span style="color:blue;font-weight:bold;">Table</span>

Classes can be organized hierarchically:
* <span style="color:blue;font-weight:bold;">CheesePizza</span> <span style="color:purple;font-family:monospace;">subClassOf</span> <span style="color:blue;font-weight:bold;">Pizza</span>  (a cheese pizza is a pizza)

Classes can be defined using logical conditions:
* <span style="color:blue;font-weight:bold;">CheesePizza</span> <span style="color:purple;font-family:monospace;">subClassOf</span> <span style="color:blue;font-weight:bold;">Pizza</span> <span style="color:purple;font-family:monospace;">and</span> (<span style="color:orange;">hasPart</span> <span style="color:purple;font-family:monospace;">some</span> <span style="color:blue;font-weight:bold;">CheeseTopping</span>)

---

### Individual ###
An <span style="color:green;font-style:italic;">individual</span> is a specific instance of a class. We use `rdf:type` (written here as <span style="color:purple;font-family:monospace;">instanceOf</span> for readability) to assert class membership.
Examples:
* <span style="color:green;font-style:italic;">Michel</span> <span style="color:purple;font-family:monospace;">instanceOf</span> <span style="color:blue;font-weight:bold;">Person</span>
* <span style="color:green;font-style:italic;">pizza</span> <span style="color:purple;font-family:monospace;">instanceOf</span> <span style="color:blue;font-weight:bold;">Pizza</span>
* <span style="color:green;font-style:italic;">cheese</span> <span style="color:purple;font-family:monospace;">instanceOf</span> <span style="color:blue;font-weight:bold;">Cheese</span>

Individuals can belong to multiple classes:
* <span style="color:green;font-style:italic;">my_pizza</span> <span style="color:purple;font-family:monospace;">instanceOf</span> <span style="color:blue;font-weight:bold;">Pizza</span> <span style="color:purple;font-family:monospace;">and</span> <span style="color:blue;font-weight:bold;">CheesePizza</span>

---

### Object Property ###
An <span style="color:orange;">object property</span> links one individual to another individual. We reuse properties from SULO such as <span style="color:orange;">hasPart</span>, <span style="color:orange;">hasDirectPart</span>, <span style="color:orange;">hasFeature</span>, and <span style="color:orange;">hasParticipant</span>.

Examples:
* <span style="color:green;font-style:italic;">pizza</span> <span style="color:orange;">hasPart</span> <span style="color:green;font-style:italic;">mozzarella</span>
* <span style="color:green;font-style:italic;">pizza</span> <span style="color:orange;">hasFeature</span> <span style="color:green;font-style:italic;">spice_level</span>

Object properties can have the following characteristics:

| Characteristic | Meaning | Pizza example |
|---|---|---|
| **Transitive** | If A→B and B→C, then A→C | <span style="color:green;font-style:italic;">pizza</span> <span style="color:orange;">hasPart</span> <span style="color:green;font-style:italic;">cheese</span> and <span style="color:green;font-style:italic;">cheese</span> <span style="color:orange;">hasPart</span> <span style="color:green;font-style:italic;">protein</span> → <span style="color:green;font-style:italic;">pizza</span> <span style="color:orange;">hasPart</span> <span style="color:green;font-style:italic;">protein</span> |
| **Symmetric** | If A→B, then B→A | <span style="color:green;font-style:italic;">pizza_wiki</span> <span style="color:orange;">owl:sameAs</span> <span style="color:green;font-style:italic;">pizza_dbpedia</span> → <span style="color:green;font-style:italic;">pizza_dbpedia</span> <span style="color:orange;">owl:sameAs</span> <span style="color:green;font-style:italic;">pizza_wiki</span> (same pizza, two data sources) |
| **Asymmetric** | If A→B, then NOT B→A | <span style="color:green;font-style:italic;">pizza</span> <span style="color:orange;">hasPart</span> <span style="color:green;font-style:italic;">mozzarella</span> → <span style="color:green;font-style:italic;">mozzarella</span> does NOT <span style="color:orange;">hasPart</span> <span style="color:green;font-style:italic;">pizza</span> |
| **Reflexive** | Everything relates to itself | every <span style="color:green;font-style:italic;">pizza</span> <span style="color:orange;">hasPart</span> itself |
| **Irreflexive** | Nothing relates to itself | a <span style="color:green;font-style:italic;">pizza</span> cannot <span style="color:orange;">hasDirectPart</span> itself |
| **Functional** | Each subject has at most one object | <span style="color:green;font-style:italic;">spice_level</span> <span style="color:orange;">hasFeature</span> at most one <span style="color:green;font-style:italic;">chili_scale</span> (unit) |
| **Inverse Functional** | Each object has at most one subject | <span style="color:green;font-style:italic;">chili_scale</span> is the <span style="color:orange;">hasFeature</span> value of at most one <span style="color:green;font-style:italic;">spice_level</span> |

---

### Data Property ###
A <span style="color:orange;">data property</span> links an individual to a **literal value** (a number, string, boolean, etc.) rather than to another individual. SULO defines <span style="color:orange;">hasValue</span> as the standard data property for recording the magnitude of a quantity.

Example:
* <span style="color:green;font-style:italic;">spice_level</span> <span style="color:orange;">hasValue</span> `7` (integer) — the spiciness of a pizza on a 1–10 scale

Data properties support only **one** characteristic:

| Characteristic | Meaning | Pizza example |
|---|---|---|
| **Functional** | Each individual has at most one value | <span style="color:green;font-style:italic;">spice_level</span> has at most one <span style="color:orange;">hasValue</span> |

> Unlike object properties, data properties cannot be transitive, symmetric, reflexive, or inverse — literals cannot be subjects of triples.

### Property Structure ###

Beyond characteristics, properties can be given structural axioms that constrain how they connect classes and relate to each other. The examples below use SULO properties: <span style="color:orange;">hasPart</span>, <span style="color:orange;">hasDirectPart</span>, <span style="color:orange;">hasFeature</span>, <span style="color:orange;">hasParticipant</span>, and <span style="color:orange;">refersTo</span>.

> **Color convention:** <span style="color:blue;font-weight:bold;">Classes</span> &nbsp;|&nbsp; <span style="color:green;font-style:italic;">Individuals</span> &nbsp;|&nbsp; <span style="color:orange;">Properties</span> &nbsp;|&nbsp; <span style="color:purple;font-family:monospace;">OWL keywords</span>

**Domain and Range**

A <span style="color:purple;font-family:monospace;">domain</span> axiom says that anything that uses a property as subject must be of a certain class.
A <span style="color:purple;font-family:monospace;">range</span> axiom says that anything that appears as the object of a property must be of a certain class.

| Axiom | Meaning | Example |
|---|---|---|
| <span style="color:orange;">hasParticipant</span> <span style="color:purple;font-family:monospace;">domain</span> <span style="color:blue;font-weight:bold;">Process</span> | only processes can have participants | if <span style="color:green;font-style:italic;">x</span> <span style="color:orange;">hasParticipant</span> <span style="color:green;font-style:italic;">y</span>, then <span style="color:green;font-style:italic;">x</span> <span style="color:purple;font-family:monospace;">instanceOf</span> <span style="color:blue;font-weight:bold;">Process</span> |
| <span style="color:orange;">hasParticipant</span> <span style="color:purple;font-family:monospace;">range</span> <span style="color:blue;font-weight:bold;">SpatialObject</span> | participants must be spatial objects | if <span style="color:green;font-style:italic;">x</span> <span style="color:orange;">hasParticipant</span> <span style="color:green;font-style:italic;">y</span>, then <span style="color:green;font-style:italic;">y</span> <span style="color:purple;font-family:monospace;">instanceOf</span> <span style="color:blue;font-weight:bold;">SpatialObject</span> |

> Domain and range are **inference triggers**, not integrity constraints. If you assert <span style="color:green;font-style:italic;">pizza_assembly</span> <span style="color:orange;">hasParticipant</span> <span style="color:green;font-style:italic;">basil</span>, the reasoner will *infer* that <span style="color:green;font-style:italic;">pizza_assembly</span> is a <span style="color:blue;font-weight:bold;">Process</span> — it will not raise an error.

**Inverse Properties**

An <span style="color:purple;font-family:monospace;">inverseOf</span> axiom links two properties so that asserting one implies the other in reverse.

| Axiom | Meaning | Example |
|---|---|---|
| <span style="color:orange;">isFeatureOf</span> <span style="color:purple;font-family:monospace;">inverseOf</span> <span style="color:orange;">hasFeature</span> | if A hasFeature B, then B isFeatureOf A | <span style="color:green;font-style:italic;">spice_level</span> <span style="color:orange;">isFeatureOf</span> <span style="color:green;font-style:italic;">my_pizza</span> is inferred from <span style="color:green;font-style:italic;">my_pizza</span> <span style="color:orange;">hasFeature</span> <span style="color:green;font-style:italic;">spice_level</span> |

**Sub-properties**

A <span style="color:purple;font-family:monospace;">subPropertyOf</span> axiom makes one property a specialisation of another. Any assertion using the sub-property is also an assertion using the super-property.

| Axiom | Meaning | Example |
|---|---|---|
| <span style="color:orange;">hasDirectPart</span> <span style="color:purple;font-family:monospace;">subPropertyOf</span> <span style="color:orange;">hasPart</span> | every direct part is also a part | <span style="color:green;font-style:italic;">my_pizza</span> <span style="color:orange;">hasDirectPart</span> <span style="color:green;font-style:italic;">dough</span> → <span style="color:green;font-style:italic;">my_pizza</span> <span style="color:orange;">hasPart</span> <span style="color:green;font-style:italic;">dough</span> |

**Property Chains**

A <span style="color:purple;font-family:monospace;">propertyChain</span> axiom infers a new relationship from a sequence of two existing ones.

| Axiom | Meaning | Example |
|---|---|---|
| <span style="color:orange;">hasParticipant</span> ∘ <span style="color:orange;">isFeatureOf</span> <span style="color:purple;font-family:monospace;">subPropertyOf</span> <span style="color:orange;">hasParticipant</span> | if a process has a role as participant, and that role is a feature of an object, then the process has that object as a participant | <span style="color:green;font-style:italic;">pizza_making</span> <span style="color:orange;">hasParticipant</span> <span style="color:green;font-style:italic;">topping_role</span> and <span style="color:green;font-style:italic;">topping_role</span> <span style="color:orange;">isFeatureOf</span> <span style="color:green;font-style:italic;">sliced_peppers</span> → <span style="color:green;font-style:italic;">pizza_making</span> <span style="color:orange;">hasParticipant</span> <span style="color:green;font-style:italic;">sliced_peppers</span> |

### OWL Class Expressions ###

A **class expression** builds a new (anonymous) class from existing classes and properties using logical constructors. Class expressions appear in restrictions and definitions. All property examples below use SULO relations.

> **Color convention:** <span style="color:blue;font-weight:bold;">Classes</span> &nbsp;|&nbsp; <span style="color:green;font-style:italic;">Individuals</span> &nbsp;|&nbsp; <span style="color:orange;">Properties</span> &nbsp;|&nbsp; <span style="color:purple;font-family:monospace;">OWL keywords</span>

| Constructor | Syntax | Meaning | Pizza example |
|---|---|---|---|
| **Intersection** | A <span style="color:purple;font-family:monospace;">and</span> B | things that are both A and B | <span style="color:blue;font-weight:bold;">Pizza</span> <span style="color:purple;font-family:monospace;">and</span> <span style="color:orange;">hasPart</span> <span style="color:purple;font-family:monospace;">some</span> <span style="color:blue;font-weight:bold;">CheeseTopping</span> |
| **Union** | A <span style="color:purple;font-family:monospace;">or</span> B | things that are A or B (or both) | <span style="color:blue;font-weight:bold;">VegetarianTopping</span> <span style="color:purple;font-family:monospace;">or</span> <span style="color:blue;font-weight:bold;">MeatTopping</span> |
| **Complement (negation)** | <span style="color:purple;font-family:monospace;">not</span> A | things that are NOT A | <span style="color:purple;font-family:monospace;">not</span> <span style="color:blue;font-weight:bold;">MeatTopping</span> |
| **Existential restriction** | P <span style="color:purple;font-family:monospace;">some</span> A | has at least one P that is an A | <span style="color:orange;">hasPart</span> <span style="color:purple;font-family:monospace;">some</span> <span style="color:blue;font-weight:bold;">SpicyTopping</span> |
| **Universal restriction** | P <span style="color:purple;font-family:monospace;">only</span> A | all P values are A (or none exist) | <span style="color:orange;">hasPart</span> <span style="color:purple;font-family:monospace;">only</span> <span style="color:blue;font-weight:bold;">VegetarianTopping</span> |
| **Min cardinality** | P <span style="color:purple;font-family:monospace;">min</span> n A | at least n P values that are A | <span style="color:orange;">hasPart</span> <span style="color:purple;font-family:monospace;">min</span> `2` <span style="color:blue;font-weight:bold;">Topping</span> |
| **Max cardinality** | P <span style="color:purple;font-family:monospace;">max</span> n A | at most n P values that are A | <span style="color:orange;">hasPart</span> <span style="color:purple;font-family:monospace;">max</span> `5` <span style="color:blue;font-weight:bold;">Topping</span> |
| **Exact cardinality** | P <span style="color:purple;font-family:monospace;">exactly</span> n A | exactly n P values that are A | <span style="color:orange;">hasDirectPart</span> <span style="color:purple;font-family:monospace;">exactly</span> `1` <span style="color:blue;font-weight:bold;">PizzaBase</span> |
| **HasValue** | P <span style="color:purple;font-family:monospace;">value</span> i | has P pointing to a specific individual i | <span style="color:orange;">hasFeature</span> <span style="color:purple;font-family:monospace;">value</span> <span style="color:green;font-style:italic;">chili_scale</span> |

Constructors compose freely. For example, **VegetarianPizza** can be *defined* as:

> <span style="color:blue;font-weight:bold;">VegetarianPizza</span> <span style="color:purple;font-family:monospace;">equivalentTo</span> <span style="color:blue;font-weight:bold;">Pizza</span> <span style="color:purple;font-family:monospace;">and</span> (<span style="color:orange;">hasPart</span> <span style="color:purple;font-family:monospace;">only</span> (<span style="color:purple;font-family:monospace;">not</span> <span style="color:blue;font-weight:bold;">MeatTopping</span>))

A pizza is vegetarian if and only if every one of its parts is not a meat topping. Using <span style="color:purple;font-family:monospace;">equivalentTo</span> (rather than <span style="color:purple;font-family:monospace;">subClassOf</span>) turns this into a *defined* class — the reasoner can automatically classify individuals into it.

### Class Axioms ###

Class axioms constrain how classes relate to each other. They are essential for enabling classification and detecting modelling errors.

> **Color convention:** <span style="color:blue;font-weight:bold;">Classes</span> &nbsp;|&nbsp; <span style="color:green;font-style:italic;">Individuals</span> &nbsp;|&nbsp; <span style="color:orange;">Properties</span> &nbsp;|&nbsp; <span style="color:purple;font-family:monospace;">OWL keywords</span>

**SubClass vs EquivalentTo**

| Axiom | Meaning | Pizza example |
|---|---|---|
| A <span style="color:purple;font-family:monospace;">subClassOf</span> B | every A is a B (necessary condition) | <span style="color:blue;font-weight:bold;">MargheritaPizza</span> <span style="color:purple;font-family:monospace;">subClassOf</span> <span style="color:blue;font-weight:bold;">Pizza</span> |
| A <span style="color:purple;font-family:monospace;">equivalentTo</span> B | A and B have exactly the same members (necessary and sufficient) | <span style="color:blue;font-weight:bold;">VegetarianPizza</span> <span style="color:purple;font-family:monospace;">equivalentTo</span> <span style="color:blue;font-weight:bold;">Pizza</span> <span style="color:purple;font-family:monospace;">and</span> (<span style="color:orange;">hasPart</span> <span style="color:purple;font-family:monospace;">only</span> (<span style="color:purple;font-family:monospace;">not</span> <span style="color:blue;font-weight:bold;">MeatTopping</span>)) |

Only <span style="color:purple;font-family:monospace;">equivalentTo</span> (defined classes) enables the reasoner to automatically *classify* individuals — with <span style="color:purple;font-family:monospace;">subClassOf</span> alone, the reasoner cannot infer membership.

**Disjoint Classes**

A <span style="color:purple;font-family:monospace;">DisjointWith</span> axiom states that two classes share no members. This lets the reasoner detect contradictions.

| Axiom | Meaning | Pizza example |
|---|---|---|
| A <span style="color:purple;font-family:monospace;">DisjointWith</span> B | nothing can be both A and B | <span style="color:blue;font-weight:bold;">MeatTopping</span> <span style="color:purple;font-family:monospace;">DisjointWith</span> <span style="color:blue;font-weight:bold;">VegetarianTopping</span> |
| <span style="color:purple;font-family:monospace;">AllDisjoint</span>(A, B, C, ...) | no two of these classes share a member | <span style="color:purple;font-family:monospace;">AllDisjoint</span>(<span style="color:blue;font-weight:bold;">CheeseTopping</span>, <span style="color:blue;font-weight:bold;">MeatTopping</span>, <span style="color:blue;font-weight:bold;">VegetableTopping</span>) |

Without disjointness, a reasoner cannot conclude inconsistency even if an individual is asserted to be both a <span style="color:blue;font-weight:bold;">MeatTopping</span> and a <span style="color:blue;font-weight:bold;">VegetarianTopping</span>.

**Covering Axioms (Closing a Class)**

A covering axiom combines <span style="color:purple;font-family:monospace;">subClassOf</span> with a union to state that a class is *exhausted* by its subclasses:

> <span style="color:blue;font-weight:bold;">Topping</span> <span style="color:purple;font-family:monospace;">subClassOf</span> (<span style="color:blue;font-weight:bold;">CheeseTopping</span> <span style="color:purple;font-family:monospace;">or</span> <span style="color:blue;font-weight:bold;">MeatTopping</span> <span style="color:purple;font-family:monospace;">or</span> <span style="color:blue;font-weight:bold;">VegetableTopping</span>)

Combined with <span style="color:purple;font-family:monospace;">AllDisjoint</span>, this closes the class: every topping is exactly one of cheese, meat, or vegetable — nothing else.

### Individual Assertions ###

**Assertions** are statements about specific individuals — the "facts" of the ontology. There are seven kinds:

> **Color convention:** <span style="color:blue;font-weight:bold;">Classes</span> &nbsp;|&nbsp; <span style="color:green;font-style:italic;">Individuals</span> &nbsp;|&nbsp; <span style="color:orange;">Properties</span> &nbsp;|&nbsp; <span style="color:purple;font-family:monospace;">OWL keywords</span>

| Assertion | Meaning | Pizza example |
|---|---|---|
| **Class assertion** | individual i is of type C | <span style="color:green;font-style:italic;">my_pizza</span> <span style="color:purple;font-family:monospace;">instanceOf</span> <span style="color:blue;font-weight:bold;">MargheritaPizza</span> |
| **Object property assertion** | i is related to j via P | <span style="color:green;font-style:italic;">my_pizza</span> <span style="color:orange;">hasPart</span> <span style="color:green;font-style:italic;">mozzarella</span> |
| **Negative object property assertion** | i is explicitly NOT related to j via P | <span style="color:purple;font-family:monospace;">not</span> (<span style="color:green;font-style:italic;">my_pizza</span> <span style="color:orange;">hasPart</span> <span style="color:green;font-style:italic;">anchovies</span>) |
| **Data property assertion** | i has literal value v for P | <span style="color:green;font-style:italic;">spice_level</span> <span style="color:orange;">hasValue</span> `7` |
| **Negative data property assertion** | i explicitly does NOT have literal value v for P | <span style="color:purple;font-family:monospace;">not</span> (<span style="color:green;font-style:italic;">spice_level</span> <span style="color:orange;">hasValue</span> `0`) |
| **Same individual** | i and j refer to the same real-world thing | <span style="color:green;font-style:italic;">my_pizza</span> <span style="color:purple;font-family:monospace;">sameAs</span> <span style="color:green;font-style:italic;">michels_pizza</span> |
| **Different individuals** | i and j are distinct things | <span style="color:green;font-style:italic;">margherita</span> <span style="color:purple;font-family:monospace;">differentFrom</span> <span style="color:green;font-style:italic;">pepperoni</span> |

#### Why negation matters

OWL operates under the **Open World Assumption (OWA)**: if something is not stated, it is *unknown* — not false. This has a critical consequence:

> Stating that <span style="color:green;font-style:italic;">my_pizza</span> <span style="color:orange;">hasPart</span> <span style="color:green;font-style:italic;">mozzarella</span> does **not** mean it has no other toppings.

To make a negative fact explicit you must use a **negative assertion**:

> <span style="color:purple;font-family:monospace;">not</span> (<span style="color:green;font-style:italic;">my_pizza</span> <span style="color:orange;">hasPart</span> <span style="color:green;font-style:italic;">anchovies</span>)

This tells the reasoner that the absence of anchovies is a known fact, not just a gap in the data. Without it, a reasoner cannot conclude the pizza is vegetarian even if no meat toppings are listed.

### Annotation Properties ###

**Annotation properties** attach human-readable metadata to ontology elements — classes, individuals, properties, or the ontology itself. They carry no logical meaning; the reasoner ignores them, but they are essential for FAIR ontologies.

> **Color convention:** <span style="color:blue;font-weight:bold;">Classes</span> &nbsp;|&nbsp; <span style="color:green;font-style:italic;">Individuals</span> &nbsp;|&nbsp; <span style="color:orange;">Properties</span> &nbsp;|&nbsp; <span style="color:purple;font-family:monospace;">OWL keywords</span>

**Common annotation properties**

| Property | Vocabulary | Purpose | Pizza example |
|---|---|---|---|
| `rdfs:label` | RDFS | human-readable name | <span style="color:blue;font-weight:bold;">MargheritaPizza</span> `rdfs:label` `"Margherita Pizza"@en` |
| `rdfs:comment` | RDFS | free-text description | <span style="color:blue;font-weight:bold;">MargheritaPizza</span> `rdfs:comment` `"A pizza with tomato and mozzarella"@en` |
| `skos:definition` | SKOS | formal, citable definition | <span style="color:blue;font-weight:bold;">Pizza</span> `skos:definition` `"An open-faced baked pie of Italian origin..."@en` |
| `skos:altLabel` | SKOS | synonyms or alternative names | <span style="color:blue;font-weight:bold;">MargheritaPizza</span> `skos:altLabel` `"Margherita"@en` |
| `skos:example` | SKOS | illustrative example | <span style="color:blue;font-weight:bold;">SpicyTopping</span> `skos:example` `"jalapeño, nduja"@en` |
| `dc:creator` | Dublin Core | who created this ontology | `owl:Ontology` `dc:creator` `"Michel Dumontier"` |
| `owl:versionInfo` | OWL | version string | `owl:Ontology` `owl:versionInfo` `"1.0.0"` |

**Language tags**

Annotation values can carry a language tag so the same term can have labels in multiple languages:

- <span style="color:blue;font-weight:bold;">Pizza</span> `rdfs:label` `"Pizza"@en`
- <span style="color:blue;font-weight:bold;">Pizza</span> `rdfs:label` `"Pizza"@it`
- <span style="color:blue;font-weight:bold;">Pizza</span> `rdfs:label` `"Pizza"@nl`

**Ontology-level annotations**

Annotations can also describe the ontology as a whole, which is required for FAIR publication:

```
owl:Ontology
    rdfs:label        "Pizza Ontology"@en
    dc:description    "An ontology of pizza types and ingredients"@en
    owl:versionIRI    <https://w3id.org/ontostart/pizza/1.0.0>
    dc:license        <https://creativecommons.org/licenses/by/4.0/>
```

> Annotation properties do **not** affect reasoning — you can annotate freely without changing the logical content of your ontology.
